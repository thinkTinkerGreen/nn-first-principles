This notebook demonstrates 4 foundational deep learning concepts with self-contained, reproducible experiments.

## 1. Activations Exist for a Reason
Without a nonlinearity, a neural network can only produce hyperplanes (straight lines in 2D). Concentric rings are not linearly separable.

```python
acc_lin, acc_relu = run_experiment_1("../assets/01_activations_boundary.png")
print(f"Single linear layer accuracy:   {acc_lin * 100:.1f}%")
print(f"One ReLU hidden layer accuracy: {acc_relu * 100:.1f}%")
```

## 2. Depth Without Nonlinearity is a Lie
Five stacked linear layers collapse algebraically into a single linear map: $\tilde{W} = W_5 W_4 W_3 W_2 W_1$. Depth alone adds zero representational power.

```python
acc1, acc5_lin, acc5_relu, max_err = run_experiment_2("../assets/02_depth_collapse.png")
print(f"1-Layer Linear accuracy: {acc1 * 100:.1f}%")
print(f"5-Layer Linear accuracy: {acc5_lin * 100:.1f}%")
print(f"5-Layer + ReLU accuracy: {acc5_relu * 100:.1f}%")
print(f"Max absolute error between 5-layer and collapsed single matrix: {max_err:.2e}")
```

## 3. Embeddings Learn Similarity from Nothing But Next-Token
Trained exclusively to predict the next token in a synthetic grammar, word vectors cluster by syntactic and semantic categories.

```python
loss, purity = run_experiment_3("../assets/03_emergent_embeddings.png")
print(f"Next-token Cross-Entropy loss: {loss:.4f}")
print(f"2-NN Semantic Category Purity:  {purity * 100:.1f}%")
```

## 4. Memorization vs. Generalization (Data Closes the Gap)
An over-parameterized model easily memorizes small datasets ($N=20$), achieving 0 training loss but failing on held-out test data. Increasing $N$ closes the gap.

```python
summary = run_experiment_4("../assets/04_generalization_gap.png")
for n, stats in summary.items():
    print(f"N={n:4d}: Train Loss={stats['final_train_loss']:.3f}, Test Loss={stats['final_test_loss']:.3f}, Gap={stats['generalization_gap']:.3f}, Test Acc={stats['final_test_acc']*100:.1f}%")
```
